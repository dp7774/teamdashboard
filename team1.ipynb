```python
beta = origin.copy()
```

```python
# 월, 수출입구분 '입항', 선종명만 추출
beta_visit = beta[
    beta['수출입구분'] == '입항'
][['월', '선종명', '전체선박수']].copy()

beta_visit
```

```python
ship_rank = (
    beta_visit
    .groupby(['월', '선종명'], as_index=False)['전체선박수']
    .sum()
)

ship_rank = ship_rank.sort_values(
    by=['월', '전체선박수'],
    ascending=[True, False]
)

ship_rank = ship_rank.reset_index(drop=True)

ship_rank
```

```python
# 1. 선종별 입항 비중

ship_share = (
    beta_visit
    .groupby('선종명', as_index=False)['전체선박수']
    .sum()
)
ship_share
```

```python
total_ships = ship_share['전체선박수'].sum()
total_ships
```

```python
ship_share['입항비중(%)'] = (
    ship_share['전체선박수']
    / total_ships
    * 100
).round(2)
ship_share

"""
선종별 입항 비중
= 해당 선종 입항 선박 수
  ÷ 전체 입항 선박 수
  × 100
"""
```

```python
# 2. 월별 입항 선박 수

monthly_visit = (
    beta_visit
    .groupby('월', as_index=False)['전체선박수']
    .sum()
)
monthly_visit
```

```python
monthly_visit = monthly_visit.sort_values(
    by='월'
).reset_index(drop=True)

monthly_visit
```

```python
# 3. 전월 대비 입항량 증감율

monthly_visit['전월대비증감률(%)'] = (
    monthly_visit['전체선박수']
    .pct_change()
    * 100
).round(2)

monthly_visit
```

```python
# 4. 총 입항 선박수

total_visit = beta_visit['전체선박수'].sum()
total_visit
```
